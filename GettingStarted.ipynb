```python
# chatprompt template
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ('system', "You are an expert AI engineer. Provide me answer based on the question and at last include Knowledge check."),
        ('user', "{input}")
    ]
)

prompt
```

```python
chain = prompt|llm
response = chain.invoke({"input" : "Explain me about LangSmith."})
print(response)
```

```python
## stroutput parser
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt|llm

response = chain.invoke({"input" : "Explain me about LangSmith."})
print(response)

response3 = output_parser.invoke(response)
print("----------------------------------------------------------------------------")
print("----------------------------------------------------------------------------")
print("----------------------------------------------------------------------------")
print("----------------------------------------------------------------------------")
print(response3)
```
